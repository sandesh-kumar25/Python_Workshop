```python
class Product:
    def __init__(self, pid, name, price, stock):
        self.id = pid
        self.name = name
        self.price = price
        self.stock = stock

    def show(self):
        return f"{self.id}: {self.name} - ₹{self.price} (Stock: {self.stock})"

class CartItem:
    def __init__(self, product, qty):
        self.product = product
        self.qty = qty

    def subtotal(self):
        return self.product.price * self.qty

class ShoppingCart:
    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.catalog = {}
        self.cart = {}

    def add_product(self, p):
        self.catalog[p.id] = p

    def view_products(self):
        print("\nAvailable Products:")
        for p in self.catalog.values():
            print(p.show())

    def add_to_cart(self, pid, qty):
        product = self.catalog.get(pid)
        if product:
            if product.stock >= qty:
                product.stock -= qty
                if pid in self.cart:
                    self.cart[pid].qty += qty
                else:
                    self.cart[pid] = CartItem(product, qty)
                print(f"{qty} x {product.name} added to cart.")
            else:
                print("Not enough stock.")
        else:
            print("Invalid product ID.")

    def remove_from_cart(self, pid, qty):
        if pid in self.cart:
            item = self.cart[pid]
            if qty >= item.qty:
                item.product.stock += item.qty
                del self.cart[pid]
                print(f"Removed all of {item.product.name} from cart.")
            else:
                item.qty -= qty
                item.product.stock += qty
                print(f"Removed {qty} of {item.product.name} from cart.")
        else:
            print("Item not in cart.")

    def display_cart(self):
        if not self.cart:
            print("\nYour cart is empty.")
            return
        print("\nYour Cart:")
        for item in self.cart.values():
            print(f"{item.product.name} x {item.qty} = ₹{item.subtotal()}")

    def edit_cart(self):
        while True:
            self.display_cart()
            choice = input("\nType 'remove' to remove item, 'back' to return: ").strip().lower()
            if choice == 'back':
                break
            elif choice == 'remove':
                pid = input("Enter Product ID to remove: ").strip()
                if pid in self.cart:
                    try:
                        qty = int(input("Enter quantity to remove: "))
                        self.remove_from_cart(pid, qty)
                    except ValueError:
                        print("Invalid quantity!")
                else:
                    print("Product not in cart.")
            else:
                print("Invalid choice!")

    def display_bill(self):
        print(f"\nFinal Bill for {self.customer_name}")
        print("-----------------------------")
        total = 0
        for item in self.cart.values():
            sub = item.subtotal()
            print(f"{item.product.name} x {item.qty} = ₹{sub}")
            total += sub
        print("-----------------------------")
        print(f"Total Amount: ₹{total}")
        print("Thank you for shopping!")
        print("Visit Again\n")

def run_shopping():
    name = input("Enter your name: ")
    shop = ShoppingCart(name)

    # Product list
    shop.add_product(Product("P1", "Rasgulla (1 box)", 250, 10))
    shop.add_product(Product("P2", "Brown Bread", 60, 15))
    shop.add_product(Product("P3", "Curd (1kg)", 77, 20))
    shop.add_product(Product("P4", "Curd (500gm)", 40, 25))
    shop.add_product(Product("P5", "Paneer (1kg)", 320, 10))
    shop.add_product(Product("P6", "White Bread", 40, 15))
    shop.add_product(Product("P7", "Peanut Butter (510g)", 320, 10))
    shop.add_product(Product("P8", "Real Mixed Fruit Juice (1L)", 220, 12))

    print(f"\nWelcome {name}! Start adding items to your cart.")

    while True:
        print("\nMenu:")
        print("1. View Products and Add to Cart")
        print("2. View and Edit Cart")
        print("3. Checkout and Generate Bill")
        choice = input("Choose an option (1/2/3): ").strip()

        if choice == '1':
            shop.view_products()
            pid = input("Enter Product ID to buy: ").strip()
            if pid not in shop.catalog:
                print("**Invalid product ID. Please try again.**\n")
                continue
            try:
                qty = int(input("Enter quantity: "))
                if qty <= 0:
                    print("Quantity must be greater than 0.\n")
                    continue
                shop.add_to_cart(pid, qty)
            except ValueError:
                print("**Invalid quantity! Please enter a number.**\n")

        elif choice == '2':
            shop.edit_cart()

        elif choice == '3':
            break
        else:
            print("Invalid choice! Please enter 1, 2, or 3.")

    shop.display_bill()

run_shopping()
```
